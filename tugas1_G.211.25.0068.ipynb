```python
def hitung_total(data):
  hasil = 0
  for i in range(len(data)):
      hasil = hasil + data[i]
  return hasil

angka = [3, 5, 1, 2]
print(hitung_total(angka))
```

```python
panjang = 3
lebar = 3
data = [(1, 7, 3),(4, 5, 8),(2, 9, 6)]
total = 0

for i in range(panjang):
    for j in range(lebar):
        total += data[i][j]
print(total)
```
