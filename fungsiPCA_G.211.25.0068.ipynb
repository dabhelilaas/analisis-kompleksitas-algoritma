```python
#fungsi algoritma pca
x = [[1, 2],
    [2, 1],
    [3, 5],
    [4, 4]]
rata = []
for i in range(len(x[0])):
    total = 0
    for j in range(len(x)):
        total = total + x[j][i]
    rata.append(total / len(x))
print("Rata-rata:")
print(rata)

hasil = []
for i in range(len(x)):
    sementara = []
    for j in range(len(x[i])):
        sementara.append(x[i][j] - rata[j])
    hasil.append(sementara)
print("\nHasil Centering:")
for i in hasil:
    print(i)

t = []
for i in range(len(hasil[0])):
    row = []
    for j in range(len(hasil)):
        row.append(hasil[j][i])
    t.append(row)

covariance = []
for i in range(len(t)):
    temp = []
    for j in range(len(hasil[0])):
        jumlah = 0
        for k in range(len(hasil)):
            jumlah += t[i][k] * hasil[k][j]
        temp.append(jumlah / (len(hasil) - 1))
    covariance.append(temp)

print("\nCovariance:")
for i in covariance:
    print(i)
```
