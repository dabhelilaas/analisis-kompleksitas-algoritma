```python
import math

def euclidean_distance(a, b):
    jumlah = 0
    for i in range(len(a)):
        jumlah += (a[i] - b[i]) ** 2
    return math.sqrt(jumlah)

#contoh
x = [1, 2]
y = [4, 6]
print(euclidean_distance(x, y))
```

```python
def manhattan_distance(a, b):
    total = 0
    for i in range(len(a)):
        total += abs(a[i] - b[i])
    return total

print (manhattan_distance([1, 2], [4, 6]))
```
