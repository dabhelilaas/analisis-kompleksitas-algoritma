PYTHON

```python
#Bubble sort
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
       for j in range(0, n - i - 1):
          if arr[j] > arr[j + 1]:
             arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr
data = [5, 2, 9, 1, 3]
print(bubble_sort(data))
```

```python
#Selection sort
def selection_sort(data):
    n = len(data)

    for i in range(n):
        posisi_min = i
        for j in range(i + 1, n):
            if data[j] < data[posisi_min]:
                posisi_min = j
        data[i], data[posisi_min] = data[posisi_min], data[i]
    return data
angka = [25, 11, 9, 18]
print("Sebelum sorting:", angka)
hasil = selection_sort(angka)
print("Sesudah sorting:", hasil)
```

```python
#Insertion sort
def insertion_sort(data):
    for i in range(1, len(data)):
        key = data[i]
        j = i - 1
        while j >= 0 and data[j] > key:
            data[j + 1] = data[j]
            j -= 1
        data[j + 1] = key
    return data
angka = [16, 5, 12, 8]
print("Data sebelum:", angka)

hasil = insertion_sort(angka)
print("Data sesudah:", hasil)
```

```python
#Merge sort
def merge_sort(data):
    if len(data) > 1:
        tengah = len(data) // 2

        kiri = data[:tengah]
        kanan = data[tengah:]

        merge_sort(kiri)
        merge_sort(kanan)

        i = j = k = 0

        while i < len(kiri) and j < len(kanan):
            if kiri[i] < kanan[j]:
                data[k] = kiri[i]
                i += 1
            else:
                data[k] = kanan[j]
                j += 1

            k += 1

        while i < len(kiri):
            data[k] = kiri[i]
            i += 1
            k += 1

        while j < len(kanan):
            data[k] = kanan[j]
            j += 1
            k += 1

    return data

angka = [30, 6, 21, 15]
print("Data awal:", angka)

hasil = merge_sort(angka)
print("Data akhir:", hasil)
```

ANALISA

```python
#Bubble Sort
Bubble Sort bekerja dengan membandingkan dua data yang berdekatan lalu menukarnya jika urutannya salah
Proses dilakukan berulang sampai semua data tersusun dari kecil ke besar.
```

```python
#Selection Sort
Selection Sort mencari nilai paling kecil dari data yang belum terurut, lalu memindahkannya ke posisi yang benar
Proses dilakukan satu per satu hingga seluruh data tersusun rapi.
```

```python
#Insertion Sort
Insertion Sort menyisipkan data ke posisi yang tepat pada bagian array yang sudah terurut
Metode ini cocok digunakan untuk jumlah data yang tidak terlalu banyak.
```

```python
#Merge Sort
Merge Sort membagi data menjadi beberapa bagian kecil, kemudian mengurutkannya dan menggabungkannya kembali secara teratur
Metode ini lebih cepat untuk data yang jumlahnya banyak.
```
