```python
Mulai

Input data kota & jarak
start = a
tujuan = b, c

min = tak hingga
best = kosong

Untuk setiap urutan tujuan:
    total = 0
    sekarang = start

    Untuk tiap kota:
        total += jarak(sekarang, kota)
        sekarang = kota

    total += jarak(sekarang, start)

    Jika total < min:
        min = total
        best = urutan

Output best, min

Selesai
```

coding

```python
from itertools import permutations

jarak = {
    ('a','b'):4, ('a','c'):2, ('b','c'):10,
    ('b','a'):4, ('c','a'):2, ('c','b'):10
}

kota = ['b','c']
start = 'a'

min = float('inf')
best = ()

for r in permutations(kota):
    total = 0
    pos = start

    for k in r:
        total += jarak[(pos, k)]
        pos = k

    total += jarak[(pos, start)]

    if total < min:
        min = total
        best = r

print("Rute:", "a ->", " -> ".join(best), "-> a")
print("Jarak:", min)
```

analisis kompleksitas algoritma

```python
Algoritma menggunakan seluruh permutasi rute untuk mencoba semua kemungkinan urutan kota sebanyak n!.
setiap permutasi rute dihitung total jaraknya dengan melakukan n langkah perhitungan.
maka kompleksitas waktu adalah o(n × n!).

t(n) = n × n!
o(n × n!)
```
